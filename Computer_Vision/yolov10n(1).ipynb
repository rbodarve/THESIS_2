```python
from ultralytics import YOLOv10

model = YOLOv10(f'{HOME}/runs/detect/train/weights/best.pt')

dataset = sv.DetectionDataset.from_yolo(
    images_directory_path=f"{dataset.location}/valid/images",
    annotations_directory_path=f"{dataset.location}/valid/labels",
    data_yaml_path=f"{dataset.location}/data.yaml"
)

bounding_box_annotator = sv.BoxAnnotator()
label_annotator = sv.LabelAnnotator()
```

```python
import random
import cv2
import supervision as sv
import time

annotated_images = []
latencies = []

start_total_time = time.time()

for _ in range(10):
    random_image_path = random.choice(list(dataset.image_paths))

    results = model(source=random_image_path, conf=0.25)[0]

    # Calculate latency for this image (preprocess + inference + postprocess)
    # speed values are in milliseconds
    latency = results.speed['preprocess'] + results.speed['inference'] + results.speed['postprocess']
    latencies.append(latency)

    detections = sv.Detections.from_ultralytics(results)

    image_to_annotate = cv2.imread(random_image_path)

    annotated_image = bounding_box_annotator.annotate(
        scene=image_to_annotate.copy(), detections=detections)
    annotated_image = label_annotator.annotate(
        scene=annotated_image, detections=detections)
    annotated_images.append(annotated_image)

end_total_time = time.time()
total_runtime = end_total_time - start_total_time

avg_latency = sum(latencies) / len(latencies)
fps = 1000 / avg_latency if avg_latency > 0 else 0

print(f"Average Latency: {avg_latency:.2f} ms")
print(f"Average FPS: {fps:.2f}")
print(f"Total Runtime: {total_runtime:.2f} seconds")

sv.plot_images_grid(annotated_images,  grid_size=(2, 5), # Specify 2 rows and 5 columns
    size=(10, 10)      # Optional: set the overall figure size in inches
                    )
```

```python
# The onnx-tf library is incompatible with Python 3.12 and modern ONNX.
# We will use onnx2tf, a modern alternative, to convert the model.
# We explicitly pin onnx==1.16.1 to avoid compatibility issues with onnx-graphsurgeon
# We also install ai-edge-litert as it is a dependency for onnx2tf verification
!pip install onnx==1.16.1 onnx2tf sng4onnx onnxsim onnx_graphsurgeon flatbuffers ai-edge-litert --upgrade -q

import os
import onnx
import subprocess

# Define paths
onnx_model_path = '/content/best.onnx'
tf_model_path = './tf_model'

if not os.path.exists(onnx_model_path):
    raise FileNotFoundError(f"ONNX model not found at {onnx_model_path}")

# Load ONNX model (for compatibility with later cells)
onnx_model = onnx.load(onnx_model_path)

# Perform conversion using onnx2tf
print("Converting ONNX to TensorFlow SavedModel using onnx2tf...")

# Use subprocess to capture output for debugging
cmd = ['onnx2tf', '-i', onnx_model_path, '-o', tf_model_path]

# Check for auto-generated replacement JSON from previous failures
json_path = os.path.join(tf_model_path, 'best_auto.json')
if os.path.exists(json_path):
    print(f"Found replacement JSON at {json_path}, using it to fix conversion errors.")
    cmd.extend(['-prf', json_path])

try:
    result = subprocess.run(cmd, check=True, capture_output=True, text=True)
    print(result.stdout)
    print(f"Successfully loaded ONNX and converted to TensorFlow SavedModel at {tf_model_path}")
except subprocess.CalledProcessError as e:
    print("STDOUT:", e.stdout)
    print("STDERR:", e.stderr)
    raise RuntimeError("onnx2tf conversion failed. See STDERR above for details.") from e

# --- Compatibility Layer for Next Cells ---
# The next cells expect 'prepare' from onnx_tf. We mock it to prevent errors.
class MockRep:
    def export_graph(self, path):
        print(f"(Skipping export: Model was already converted to {path} by onnx2tf)")

mock_rep = MockRep()

def prepare(model):
    return mock_rep
```
